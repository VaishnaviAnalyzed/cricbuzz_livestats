# Cricbuzz LiveStats - data fetching playground
Test API calls and push results into the SQL database.
Run this notebook from the `notebooks/` folder with the project's virtualenv.

```python
import sys
sys.path.append('..')  # make project modules importable
import pandas as pd
from utils.api_client import CricbuzzClient, flatten_matches, parse_topstats
from utils.api_to_db import save_matches
from utils.db_connection import init_db, run_query

init_db(seed=True)  # create tables + sample data if empty
client = CricbuzzClient()
print('API key configured:', client.configured)
```

## 1. Fetch recent matches

```python
payload = client.matches('recent')
matches = flatten_matches(payload)
pd.DataFrame(matches).head()
```

## 2. Save them into the database

```python
saved, skipped = save_matches(matches)
print(f'saved={saved}, skipped={skipped}')
```

## 3. Top stats from the API

```python
parse_topstats(client.top_stats('mostRuns')).head(10)
```

## 4. Query the database

```python
run_query('SELECT match_date, description, status FROM matches ORDER BY match_id DESC LIMIT 10')
```
